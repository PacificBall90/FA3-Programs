```python
#FA3-Prog1
student_name = input("Please enter your name: ")
student_age = input("Please enter your age: ")
student_course = input("Please enter your course: ")

print("")
if student_name and student_age and student_course:
  print("STUDENT INFORMATION")
  print(f"Your name is {student_name}")
  print(f"You are {student_age} years old")
  print(f"You took a course in {student_course}")
else:
  print("ERROR: Entered incorrect data")

```

```python
#FA3-Prog3
student_name = input("Please enter the student's name: ")
student_score = int(input("Please enter the student's score: "))

print()
if student_name and student_score:
  print(f"{student_name} - {student_score}")
  if student_score >= 90 and student_score <= 100:
    print("A")
  elif student_score >= 80 and student_score < 90:
    print("B")
  elif student_score >= 70 and student_score < 80:
    print("C")
  elif student_score < 70:
    print("F")
else:
  print("ERROR: Entered incorrect data")
```
